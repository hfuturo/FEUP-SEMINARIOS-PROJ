Here is a table with the information summarized.

| Feature | Meaning | Values |
|---------|---------|--------|
| **having_IP_Address** | URL uses IP instead of domain | `1 = no`, `-1 = yes` |
| **URL_Length** | Length of URL | `1 = short (<54)`, `0 = suspicious (54–75)`, `-1 = very long (>75)` |
| **Shortining_Service** | Use of URL shortener (bit.ly, tinyurl) | `1 = no`, `-1 = yes` |
| **having_At_Symbol** | Presence of `@` in URL | `1 = no`, `-1 = yes` |
| **double_slash_redirecting** | Extra `//` beyond protocol | `1 = no`, `-1 = yes` |
| **Prefix_Suffix** | Dash `-` in domain name | `1 = no`, `-1 = yes` |
| **having_Sub_Domain** | Number of subdomains | `1 = low (≤2)`, `0 = suspicious (=3)`, `-1 = high (>3)` |
| **SSLfinal_State** | HTTPS certificate validity | `1 = trusted (≥2 yrs)`, `0 = suspicious`, `-1 = none/fake` |
| **Domain_registeration_length** | Domain validity duration | `1 = ≥1 year`, `-1 = <1 year` |
| **Favicon** | Favicon loaded from external domain | `1 = same domain`, `-1 = different` |
| **port** | Use of unusual ports | `1 = normal`, `-1 = suspicious` |
| **HTTPS_token** | “https” in domain part of URL | `1 = no`, `-1 = yes` |
| **Request_URL** | % of objects from external domains | `1 = <20%`, `0 = 20–50%`, `-1 = >50%` |
| **URL_of_Anchor** | Anchors pointing outside or empty | `1 = <20%`, `0 = 20–50%`, `-1 = >50%` |
| **Links_in_tags** | Meta/script/link tags from other domains | `1 = <20%`, `0 = 20–50%`, `-1 = >50%` |
| **SFH (Server Form Handler)** | Where form data is submitted | `1 = same domain`, `0 = different domain`, `-1 = empty/about:blank` |
| **Submitting_to_email** | Form submits to email address | `1 = no`, `-1 = yes` |
| **Abnormal_URL** | Hostname doesn’t match WHOIS identity | `1 = no`, `-1 = yes` |
| **Redirect** | Number of redirections | `0 = ≤1`, `1 = >1` |
| **on_mouseover** | JavaScript changes status bar | `1 = no`, `-1 = yes` |
| **RightClick** | Right click disabled | `1 = no`, `-1 = yes` |
| **popUpWidnow** | Credentials requested in popup | `1 = no`, `-1 = yes` |
| **Iframe** | Use of `<iframe>` | `1 = no`, `-1 = yes` |
| **age_of_domain** | Age of domain | `1 = ≥6 months`, `-1 = <6 months` |
| **DNSRecord** | Existence of DNS record | `1 = found`, `-1 = not found` |
| **web_traffic** | Website popularity (Alexa rank) | `1 = top 100k`, `0 = >100k`, `-1 = none` |
| **Page_Rank** | Google PageRank | `1 = high`, `-1 = low` |
| **Google_Index** | Indexed by Google | `1 = yes`, `-1 = no` |
| **Links_pointing_to_page** | Number of backlinks | `1 = many (>2)`, `0 = some (=2)`, `-1 = few (≤1)` |
| **Statistical_report** | Flagged by blacklists | `1 = no`, `-1 = yes` |
| **Result** | Final class | `1 = legitimate`, `-1 = phishing` |


```python
import pandas as pd
```

```python
with open("dataset.arff", "r") as f:
    lines = f.readlines()

attributes = []
data_start = False
data_rows = []

for line in lines:
    line = line.strip()
    if not line or line.startswith("%"):
        continue
    if line.lower().startswith("@attribute"):
        attributes.append(line.split()[1])
    elif line.lower().startswith("@data"):
        data_start = True
    elif data_start:
        data_rows.append(line.split(","))

df = pd.DataFrame(data_rows, columns=attributes)
df.to_csv("phishing.csv", index=False)
df.head()
```
